```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from tkinter import Tk, filedialog

from sklearn.model_selection import TimeSeriesSplit, RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier

print("Library berhasil diimpor!")
```

```python
# Import library tkinter
from tkinter import Tk, filedialog

# Membuat jendela pemilihan file
root = Tk()
root.withdraw()
root.attributes('-topmost', True)

# Memilih dataset CSV
lokasi_file = filedialog.askopenfilename(
    title="Pilih Dataset Cuaca Lamongan",
    filetypes=[("CSV Files", "*.csv")]
)

root.destroy()

# Membaca dataset
if lokasi_file:
    data_asli = pd.read_csv(lokasi_file)

    print("Dataset berhasil dibaca!")
    print("Lokasi:", lokasi_file)
    print("Ukuran dataset:", data_asli.shape)

    display(data_asli.head())
else:
    raise FileNotFoundError("Dataset belum dipilih!")
```

```python
# Menyalin dataset asli
data = data_asli.copy()

# Menampilkan ukuran dataset
print("Ukuran dataset:", data.shape)

# Menampilkan lima baris pertama
display(data.head())
```

```python
# Menampilkan informasi dataset
data.info()
```

```python
# Memeriksa missing values
print("Jumlah missing values:")
print(data.isnull().sum())

# Memeriksa data duplikat
print("\nJumlah data duplikat:")
print(data.duplicated().sum())
```

```python
# Mengubah kolom tanggal menjadi datetime
data['tanggal'] = pd.to_datetime(
    data['tanggal'],
    format='%Y-%m-%d'
)

# Mengurutkan data berdasarkan tanggal
data = data.sort_values(
    by='tanggal'
).reset_index(drop=True)

# Menampilkan rentang tanggal
print("Tanggal awal:", data['tanggal'].min())
print("Tanggal akhir:", data['tanggal'].max())

display(data.head())
```

```python
# Menentukan fitur cuaca
fitur_dasar = [
    'suhu_maks (°C)',
    'suhu_min (°C)',
    'suhu_rata (°C)',
    'kelembapan_rata (%)',
    'curah_hujan (mm)',
    'durasi_hujan (jam)',
    'kecepatan_angin_maks (km/jam)',
    'hembusan_angin_maks (km/jam)',
    'radiasi_matahari (MJ/m²)'
]

print("Jumlah fitur dasar:", len(fitur_dasar))
```

```python
# Membuat salinan data untuk pemodelan
data_model = data.copy()

# Mengubah kolom fitur menjadi numerik
for kolom in fitur_dasar:
    data_model[kolom] = pd.to_numeric(
        data_model[kolom],
        errors='coerce'
    )

# Membuat fitur lag 1, 2, dan 3 hari
for lag in range(1, 4):
    for kolom in fitur_dasar:

        nama_fitur = f'{kolom}_lag_{lag}'

        data_model[nama_fitur] = data_model[kolom].shift(lag)

print("Fitur lag berhasil dibuat.")

display(data_model.head(10))
```

```python
# Membuat target cuaca satu hari berikutnya
data_model['target_cuaca_besok'] = (
    data_model['cuaca'].shift(-1)
)

# Menampilkan contoh target
display(
    data_model[
        ['tanggal', 'cuaca', 'target_cuaca_besok']
    ].head(10)
)
```

```python
# Membuat daftar seluruh fitur
fitur = fitur_dasar.copy()

# Menambahkan fitur lag
for lag in range(1, 4):
    for kolom in fitur_dasar:
        fitur.append(f'{kolom}_lag_{lag}')

# Menghapus baris yang memiliki nilai kosong
data_model = data_model.dropna(
    subset=fitur + ['target_cuaca_besok']
).reset_index(drop=True)

print("Ukuran data setelah preprocessing:", data_model.shape)
print("Jumlah fitur:", len(fitur))

display(data_model.head())
```

```python
# Membentuk variabel fitur
X = data_model[fitur]

# Membentuk variabel target
y = data_model['target_cuaca_besok']

print("Ukuran fitur X:", X.shape)
print("Ukuran target y:", y.shape)

display(X.head())
display(y.head())
```

```python
# Menentukan batas pembagian data
batas = int(len(data_model) * 0.8)

# Data training
X_train = X.iloc[:batas].copy()
y_train = y.iloc[:batas].copy()

# Data testing
X_test = X.iloc[batas:].copy()
y_test = y.iloc[batas:].copy()

# Menampilkan jumlah data
print("Jumlah data training:", len(X_train))
print("Jumlah data testing:", len(X_test))

print("\nPeriode training:")
print(
    data_model['tanggal'].iloc[0],
    "sampai",
    data_model['tanggal'].iloc[batas - 1]
)

print("\nPeriode testing:")
print(
    data_model['tanggal'].iloc[batas],
    "sampai",
    data_model['tanggal'].iloc[-1]
)
```

```python
# Distribusi kelas training
print("Distribusi kelas training:")
print(y_train.value_counts())

# Distribusi kelas testing
print("\nDistribusi kelas testing:")
print(y_test.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

# Membuat model Random Forest
model_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced'
)

# Melatih model menggunakan data training
model_rf.fit(X_train, y_train)

print("Model Random Forest berhasil dilatih!")
```

```python
# Melakukan prediksi
y_pred = model_rf.predict(X_test)

# Membuat tabel hasil prediksi
hasil_prediksi = pd.DataFrame({
    'Cuaca Aktual': y_test.values,
    'Cuaca Prediksi': y_pred
})

display(hasil_prediksi.head(20))
```

```python
# Mengambil nilai feature importance
importance = model_rf.feature_importances_

# Membuat dataframe feature importance
hasil_importance = pd.DataFrame({
    'Fitur': fitur,
    'Importance': importance
})

# Mengurutkan berdasarkan nilai importance
hasil_importance = hasil_importance.sort_values(
    by='Importance',
    ascending=False
)

# Menampilkan 15 fitur terpenting
display(hasil_importance.head(15))
```

```python
# Menyimpan hasil prediksi
hasil_akhir = data_model.iloc[batas:][
    ['tanggal', 'cuaca', 'target_cuaca_besok']
].copy()

hasil_akhir['Cuaca Aktual'] = y_test.values
hasil_akhir['Cuaca Prediksi'] = y_pred

hasil_akhir.to_csv(
    'hasil_prediksi_random_forest.csv',
    index=False
)

print("Hasil prediksi berhasil disimpan!")
```

```python
# Mengubah kategori cuaca hari ini menjadi One-Hot Encoding
cuaca_encoded = pd.get_dummies(
    data_model['cuaca'],
    prefix='cuaca_hari_ini',
    dtype=int
)

# Menggabungkan fitur numerik dan kategori cuaca
X_baru = pd.concat(
    [
        data_model[fitur].reset_index(drop=True),
        cuaca_encoded.reset_index(drop=True)
    ],
    axis=1
)

# Menentukan target cuaca besok
y_baru = data_model['target_cuaca_besok']

print("Jumlah fitur sebelumnya:", len(fitur))
print("Jumlah fitur setelah ditambahkan:", X_baru.shape[1])

display(X_baru.head())
```

```python
# Membagi data secara kronologis 80:20
batas = int(len(data_model) * 0.8)

X_train = X_baru.iloc[:batas].copy()
X_test = X_baru.iloc[batas:].copy()

y_train = y_baru.iloc[:batas].copy()
y_test = y_baru.iloc[batas:].copy()

print("Data training:", X_train.shape)
print("Data testing:", X_test.shape)
```

```python
param_grid_baru = {
    'n_estimators': [200, 300, 400, 500],
    'max_depth': [10, 15, 20, 25, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 0.8],
    'class_weight': [None, 'balanced']
}

tscv = TimeSeriesSplit(n_splits=3)

rf_baru = RandomForestClassifier(
    random_state=42,
    n_jobs=1
)

random_search_baru = RandomizedSearchCV(
    estimator=rf_baru,
    param_distributions=param_grid_baru,
    n_iter=15,
    cv=tscv,
    scoring='accuracy',
    random_state=42,
    n_jobs=1,
    verbose=1,
    pre_dispatch=1
)

# Optimasi menggunakan data training
random_search_baru.fit(X_train, y_train)

print("Parameter terbaik:")
print(random_search_baru.best_params_)

print(
    f"Akurasi validasi: "
    f"{random_search_baru.best_score_ * 100:.2f}%"
)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Mengambil model terbaik
model_rf_baru = random_search_baru.best_estimator_

# Melakukan prediksi
y_pred_baru = model_rf_baru.predict(X_test)

# Menghitung akurasi
akurasi_baru = accuracy_score(
    y_test,
    y_pred_baru
)

print(
    f"Akurasi Random Forest terbaru: "
    f"{akurasi_baru * 100:.2f}%"
)

# Classification report
print(
    classification_report(
        y_test,
        y_pred_baru,
        zero_division=0
    )
)
```
